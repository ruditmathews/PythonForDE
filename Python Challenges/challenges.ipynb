```python
import pandas as pd

# Create the DataFrame
data = {
    "project_id": [101, 101, 101, 102, 102, 103, 103],
    "member_id":  ["A",  "A",  "B",  "C",  "C",  "D",  "E"],
    "score":      [80,   90,   90,   75,   85,   95,   100]
}

df = pd.DataFrame(data)

print(df)
```

```python
df
```

```python
df1 = df.drop_duplicates(subset=["project_id","member_id"])
df1
```

```python
member_counts = df1.groupby("project_id")["member_id"].nunique()
member_counts
```

```python
valid = member_counts[member_counts>1].index
valid
```

```python
df1
```

```python
df1 = df1[df1["project_id"].isin(valid)]
df1

```

```python

df1 = df1.groupby("project_id")["score"].mean().reset_index(name = "avg_score")
df1
```
