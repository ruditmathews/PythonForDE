```python
class Circle(object):
    pi = 3.14
    def __init__(self,radius =1):
        self.radius = radius
    
    def area(self):
        return self.radius*self.radius*Circle.pi
    
    def setRadius(self,radius):
        self.radius = radius
    
    def getRadius(self):
        return self.radius
```

```python
c = Circle()
c.setRadius(2)
c.area()
```

```python
df_left = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
}, index=['K0', 'K1', 'K2'])

df_right = pd.DataFrame({
    'C': ['C0', 'C1', 'C2'],
    'D': ['D0', 'D1', 'D2']
}, index=['K0', 'K2', 'K3'])
```
